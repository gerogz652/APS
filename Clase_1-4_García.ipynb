```python
import numpy as np

import matplotlib.pyplot as plt

N = 8
n = np.arange(N)
x = 4 + 3 * np.sin((np.pi / 2) * n)

#DFT con numpy.fft 
X = np.fft.fft(x)
 
X_abs  = np.abs(X)
X_fase = np.angle(X)

plt.figure(1)
plt.title('DFT X[n]')
plt.subplot(2, 1, 1)
plt.stem(n, X_abs, basefmt=" ")
plt.grid(True)
plt.xlabel('k')
plt.ylabel('|X[k]|')

plt.subplot(2, 1, 2)
plt.stem(n, X_fase, basefmt=" ")
plt.grid(True)
plt.xlabel('k')
plt.ylabel('fase (rad)')

plt.tight_layout()
plt.show()

#%% Fase de 3/2

N = 8
n = np.arange(N)
x = 4 + 3 * np.sin((3*np.pi / 2) * n)

#DFT con numpy.fft 
X = np.fft.fft(x)
 
X_abs  = np.abs(X)
X_fase = np.angle(X)

plt.figure(2)
plt.title('DFT X[n]')
plt.subplot(2, 1, 1)
plt.stem(n, X_abs, basefmt=" ")
plt.grid(True)
plt.xlabel('k')
plt.ylabel('|X[k]|')

plt.subplot(2, 1, 2)
plt.stem(n, X_fase, basefmt=" ")
plt.grid(True)
plt.xlabel('k')
plt.ylabel('fase (rad)')

plt.tight_layout()
plt.show()

#%% Espectro de potencia

X_pot= ((X_abs)**2)/(N**2)

plt.figure(3)
plt.title('DFT X[n]')
plt.stem(n, X_pot, basefmt=" ")
plt.grid(True)
plt.xlabel('k')
plt.ylabel('|X[k]|^2')

#%% Fase de 1/3

N = 8
n = np.arange(N)
x = 4 + 3 * np.sin((1/3*np.pi) * n)

#DFT con numpy.fft 
X = np.fft.fft(x)
 
X_abs  = np.abs(X)
X_fase = np.angle(X)

plt.figure(4)
plt.title('DFT X[n]')
#plt.subplot(2, 1, 1)
plt.stem(n, X_abs, basefmt=" ")
plt.grid(True)
plt.xlabel('k')
plt.ylabel('|X[k]|')

plt.figure(5)
#plt.subplot(2, 1, 2)
plt.stem(n, X_fase, basefmt=" ")
plt.grid(True)
plt.xlabel('k')
plt.ylabel('fase (rad)')

plt.tight_layout()
plt.show()

#%% Espectro de potencia

X_pot= ((X_abs)**2)/(N**2)

plt.figure(6)
plt.title('Espectro de potencia')
plt.stem(n, X_pot, basefmt=" ")
plt.grid(True)
plt.xlabel('k')
plt.ylabel('|X[k]|^2')
```
